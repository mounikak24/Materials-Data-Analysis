```python
# ============================================================
# Cu–Zn Alloy | AI-Generated Dataset Analysis
# ChatGPT | Gemini | Perplexity | Copilot
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------
# Global plotting settings
# ---------------------------
sns.set_theme(style="whitegrid", font="sans-serif")
plt.rcParams["figure.figsize"] = (8, 5)
plt.rcParams["axes.edgecolor"] = "black"
plt.rcParams["axes.linewidth"] = 1.2

# ---------------------------
# Dataset file names
# ---------------------------
datasets = {
    "ChatGPT": "cu_zn_ai_generated_70_dataset.csv",
    "Gemini": "cu_zn_gemini_converted_70.csv",
    "Perplexity": "Perplexity_CuZn_FINAL_75rows (1).csv",
    "Copilot": "coplite_dataset (1).csv"
}

dataframes = []

# ---------------------------
# Load & clean datasets
# ---------------------------
for source, file in datasets.items():
    df = pd.read_csv(file)
    df["Source"] = source

    num_cols = df.select_dtypes(include="number").columns
    df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
    df[num_cols] = df[num_cols].fillna(df[num_cols].median())

    dataframes.append(df)

combined_df = pd.concat(dataframes, ignore_index=True)

# ---------------------------
# DISPLAY CLEANED TABLE
# ---------------------------
print("CLEANED DATA (FIRST 10 ROWS)")
display(combined_df.head(10))

# ---------------------------
# BASIC STATISTICS TABLE
# ---------------------------
print("STATISTICAL SUMMARY")
display(combined_df.describe())

# ============================================================
# HISTOGRAMS – DIFFERENT COLORS (NO YELLOW)
# ============================================================
hist_colors = [
    "navy",
    "darkgreen",
    "maroon",
    "purple",
    "teal",
    "darkorange"
]

features = [
    "Yield_Strength_MPa",
    "Ultimate_Tensile_Strength_MPa",
    "Elastic_Modulus_GPa",
    "Hardness_Value",
    "Elongation_percent",
    "Grain_Size_micrometer"
]

for col, c in zip(features, hist_colors):
    plt.figure()
    plt.hist(
        combined_df[col],
        bins=20,
        color=c,
        edgecolor="black",
        alpha=0.85
    )
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()

# ============================================================
# BOXPLOTS – AI-WISE DISTINCT COLORS
# ============================================================
box_palette = {
    "ChatGPT": "#1f77b4",
    "Gemini": "#2ca02c",
    "Perplexity": "#d62728",
    "Copilot": "#9467bd"
}

for col in features:
    plt.figure()
    sns.boxplot(
        data=combined_df,
        x="Source",
        y=col,
        palette=box_palette
    )
    plt.title(f"{col} by AI Source")
    plt.xlabel("AI Tool")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()

# ============================================================
# SCATTER PLOTS – CLEAR CONTRAST COLORS
# ============================================================

plt.figure()
sns.scatterplot(
    data=combined_df,
    x="Cu_wt_percent",
    y="Yield_Strength_MPa",
    hue="Source",
    palette=box_palette,
    s=70,
    edgecolor="black"
)
plt.title("Cu Content vs Yield Strength")
plt.xlabel("Cu (wt%)")
plt.ylabel("Yield Strength (MPa)")
plt.tight_layout()
plt.show()

plt.figure()
sns.scatterplot(
    data=combined_df,
    x="Grain_Size_micrometer",
    y="Ultimate_Tensile_Strength_MPa",
    hue="Source",
    palette=box_palette,
    s=70,
    edgecolor="black"
)
plt.title("Grain Size vs Ultimate Tensile Strength")
plt.xlabel("Grain Size (µm)")
plt.ylabel("UTS (MPa)")
plt.tight_layout()
plt.show()

# ============================================================
# SAVE FINAL CLEANED DATASET
# ============================================================
combined_df.to_csv("CuZn_ALL_AI_CLEANED_FINAL.csv", index=False)

print("✔ ALL ANALYSIS COMPLETE")
print("✔ CLEANED DATA SAVED AS: CuZn_ALL_AI_CLEANED_FINAL.csv")
```
